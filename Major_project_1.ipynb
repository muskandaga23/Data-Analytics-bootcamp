```python

#Major project 1
#atm withdrawal
import sqlite3

conn = sqlite3.connect("atm.db")
cur = conn.cursor()

cur.execute('''
CREATE TABLE IF NOT EXISTS accounts (
    acc_no INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    pin TEXT NOT NULL,
    balance REAL DEFAULT 0.0
)
''')
conn.commit()

def create_account():
    name = "Muskan"
    print(f"Creating account for: {name}")
    pin = input("Set your 4-digit PIN: ")
    deposit = float(input("Enter initial deposit amount: ₹"))

    cur.execute("INSERT INTO accounts (name, pin, balance) VALUES (?, ?, ?)", (name, pin, deposit))
    conn.commit()
    print(f"✅ Account for {name} created successfully!")

def login():
    acc_no = input("Enter your Account Number: ")
    pin = input("Enter your PIN: ")

    cur.execute("SELECT * FROM accounts WHERE acc_no = ? AND pin = ?", (acc_no, pin))
    user = cur.fetchone()

    if user:
        print(f"\nWelcome {user[1]}! 🏦")
        atm_menu(user[0])
    else:
        print("❌ Invalid account number or PIN.")

def check_balance(acc_no):
    cur.execute("SELECT balance FROM accounts WHERE acc_no = ?", (acc_no,))
    balance = cur.fetchone()[0]
    print(f"💰 Your current balance is: ₹{balance:.2f}")

def withdraw_money(acc_no):
    amount = float(input("Enter amount to withdraw: ₹"))
    cur.execute("SELECT balance FROM accounts WHERE acc_no = ?", (acc_no,))
    balance = cur.fetchone()[0]

    if amount <= balance:
        new_balance = balance - amount
        cur.execute("UPDATE accounts SET balance = ? WHERE acc_no = ?", (new_balance, acc_no))
        conn.commit()
        print(f"✅ Withdrawal successful! Remaining Balance: ₹{new_balance:.2f}")
    else:
        print("❌ Insufficient balance.")

def deposit_money(acc_no):
    amount = float(input("Enter amount to deposit: ₹"))
    cur.execute("SELECT balance FROM accounts WHERE acc_no = ?", (acc_no,))
    balance = cur.fetchone()[0]

    new_balance = balance + amount
    cur.execute("UPDATE accounts SET balance = ? WHERE acc_no = ?", (new_balance, acc_no))
    conn.commit()
    print(f"✅ Deposit successful! Updated Balance: ₹{new_balance:.2f}")

def atm_menu(acc_no):
    while True:
        print("\n====== ATM MENU ======")
        print("1. Check Balance")
        print("2. Withdraw Money")
        print("3. Deposit Money")
        print("4. Exit")
        choice = input("Select an option: ")

        if choice == "1":
            check_balance(acc_no)
        elif choice == "2":
            withdraw_money(acc_no)
        elif choice == "3":
            deposit_money(acc_no)
        elif choice == "4":
            print("👋 Thank you for using the ATM. Goodbye!")
            break
        else:
            print("❌ Invalid choice. Try again.")

while True:
    print("\n===== Welcome to ATM System =====")
    print("1. Create New Account (Name: Muskan)")
    print("2. Login to ATM")
    print("3. Exit")
    main_choice = input("Enter your choice: ")

    if main_choice == "1":
        create_account()
    elif main_choice == "2":
        login()
    elif main_choice == "3":
        print("🔒 Exiting. Have a great day!")
        break
    else:
        print("❌ Invalid input. Try again.")
conn.close()
```
